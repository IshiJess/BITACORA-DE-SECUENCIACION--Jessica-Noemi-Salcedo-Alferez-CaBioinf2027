### 5. ***Evaluación del Ensamble (QUAST)***

Para analizar la calidad del ensamble podemos usar quast desde la misma carpeta de Spades

```bash
quast --split-scaffolds -t 1 scaffolds.fasta
```

Los resultados estarán en la carpeta de Spades. Para ver los resultados hay que bajar el archivo report.html a nuestra computadora. los demás archivos también pueden bajarse, son archivos de texto y pdf.
